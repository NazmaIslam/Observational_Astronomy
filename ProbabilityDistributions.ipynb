**Jupyter Notebook to create probability distribution curves from various probability distribution**

```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

**Binomial Distribution:**
Define a Binomial Distribution

```python
def binomial_probability(n, k, p):
    """
    Calculate the probability of getting exactly k successes
    in n independent trials, where p is the probability of success.
    """
    return math.comb(n, k) * (p**k) * ((1-p)**(n-k))
```

Calculate probability of getting exactly k=3 successes in n=10 independent trials with probability of success in each trial is p=0.2

```python
n=10
k=3
p=0.2
prob_binomial = binomial_probability(n,k,p)
print(prob_binomial)
```

Plot a histogram of probability distribution of binomial distribution of n=10 and p=0.2

```python
n=10
p=0.2
k = np.arange(0, n + 1)

prob_binomial = np.array([binomial_probability(n,i,p)
    for i in k
])

# Plot probability distribution
plt.bar(k, prob_binomial, width=0.7)

plt.xlabel("Number of successes (k)")
plt.ylabel("Probability P(X = k)")
plt.title(f"Binomial Probability Distribution (n={n}, p={p})")

plt.xticks(k)
plt.grid(axis="y", alpha=0.3)

plt.show()
```

**Poisson Distribution:** Define a Poisson Distribution

```python
def poisson_probability(k1, lam):
    """
    Calculate the probability of observing exactly k events
    when the mean event rate is lambda.
    """
    return (lam**k1 * math.exp(-lam)) / math.factorial(k1)
```

Calculate probability of getting exactly k=3 for n mean value $\lambda$ of 5

```python
k1=3
lam=5
prob_poisson = poisson_probability(k1,lam)
print(prob_poisson)
```

Plot a histogram of probability distribution of poisson distribution of $\lambda$ = 5

```python
lam = 5

k1 = np.arange(0, 16)

prob_poisson = np.array([
    poisson_probability(i1, lam)
    for i1 in k1
])

plt.bar(k1, prob_poisson)

plt.xlabel("Number of events, k1")
plt.ylabel("Probability, P(X = k1)")
plt.title(f"Poisson Distribution ($\\lambda$ = {lam})")

plt.xticks(k1)
plt.grid(axis="y", alpha=0.3)

plt.show()
```

**Gaussian Distribution:** Define a Gaussian Distribution

```python
def gaussian(x, mu, sigma):
    return (1 / (sigma * np.sqrt(2 * np.pi))) * \
           np.exp(-0.5 * ((x - mu) / sigma)**2)
```

Calculate probability density at x=1 for $\mu$ = 0 and $\sigma$ = 1

```python
mu = 0
sigma = 1
x = 1

prob_gaus = gaussian(x, mu, sigma)

print(prob_gaus)
```

Plot the gaussian distribution of $\mu$ = 0 and $\sigma$ = 1

```python
mu = 0
sigma = 1

x = np.linspace(-5, 5, 1000)
y = gaussian(x, mu, sigma)

plt.plot(x, y)

plt.xlabel("$x$")
plt.ylabel("$f(x)$")
plt.title(r"Gaussian Distribution ($\mu=0$, $\sigma=1$)")

plt.grid(alpha=0.3)
plt.show()
```
